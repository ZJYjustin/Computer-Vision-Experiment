# 实验一：环境检查
补充的检查代码，不代表已在当前设备运行；原报告中的版本和 GPU 结论保留原文。

```python
import sys
import importlib
print('Python:', sys.version)
for name in ['cv2', 'numpy', 'sklearn', 'torch']:
    try:
        module = importlib.import_module(name)
        print(name, getattr(module, '__version__', 'unknown'))
        if name == 'torch':
            print('CUDA available:', module.cuda.is_available())
            print('cuDNN available:', module.backends.cudnn.is_available())
    except ImportError:
        print(name, '未安装')
```
